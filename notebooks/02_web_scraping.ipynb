### Optional schema-aware HTML extraction
The collector selects tables by flight, payload, and launch column names rather than a table index. Upstream schema errors fail explicitly. Wikipedia and the normalized API-derived snapshot have different grains and scopes and are not joined without reconciliation.

```python
from falcon9.collection import collect_wikipedia
REFRESH_WIKIPEDIA = False
if REFRESH_WIKIPEDIA:
    tables = collect_wikipedia()
    target = ROOT / 'data/live'
    target.mkdir(parents=True, exist_ok=True)
    for number, table in enumerate(tables):
        table.to_csv(target / f'wikipedia_table_{number}.csv', index=False)
    print('Extracted tables:', len(tables))
    display(tables[0].head())
else:
    print('Live HTML collection: NOT RUN. Course reference CSV inspection only.')
```
